# Text Generation vs Chat Models: Key Differences

## **Text Generation Models**
- **Purpose**: Complete or continue text based on a prompt
- **Input**: Raw text string
- **Output**: Generated text continuation
- **Use Cases**: 
  - Content creation
  - Text completion
  - Creative writing
  - Code generation
- **Example**: Input "The weather today is" → Output "sunny and warm with a gentle breeze"

## **Chat Models** 
- **Purpose**: Engage in conversational dialogue
- **Input**: Structured messages (system, human, AI roles)
- **Output**: Conversational responses
- **Use Cases**:
  - Customer support
  - Virtual assistants
  - Q&A systems
  - Interactive applications
- **Example**: Input "What's the weather like?" → Output "I don't have access to current weather data, but I can help you find weather information if you tell me your location."

## **Technical Differences**

| Aspect | Text Generation | Chat Models |
|--------|----------------|-------------|
| **Training** | Trained on text completion tasks | Trained on conversational data with reinforcement learning |
| **Context Handling** | Simple text continuation | Maintains conversation context and roles |
| **Response Style** | Completes thoughts naturally | Provides helpful, conversational responses |
| **Input Format** | Plain text strings | Structured message format |
| **Memory** | Limited context window | Better at maintaining conversation flow |

## **When to Use Which**

### Use **Text Generation** for:
- ✅ Content creation and writing assistance
- ✅ Text completion tasks
- ✅ Creative writing prompts
- ✅ Code generation
- ✅ Simple text processing

### Use **Chat Models** for:
- ✅ Interactive applications
- ✅ Question-answering systems
- ✅ Customer service bots
- ✅ Virtual assistants
- ✅ Multi-turn conversations

```python
import os
from dotenv import load_dotenv

load_dotenv("./.env")

HF_TOKEN = os.getenv("HF_TOKEN")
```

```python
from langchain_huggingface.chat_models import ChatHuggingFace
from langchain_huggingface import HuggingFaceEndpoint
from langchain_core.messages import HumanMessage, SystemMessage
from typing import List, Dict
import re

class PropertyListingEnhancer:
    """
    A class to enhance property listings using ChatHuggingFace models with fallback.
    """
    
    def __init__(self, model_repo: str = "Qwen/Qwen3-8B", temperature: float = 0.1):
        """
        Initialize the PropertyListingEnhancer.
        
        Args:
            model_repo (str): HuggingFace model repository ID
            temperature (float): Temperature for text generation
        """
        self.model_repo = model_repo
        self.temperature = temperature
        self.hf_token = HF_TOKEN
        self.provider = "featherless-ai"
        self.llm = None
        self._setup_model()
    
    def _setup_model(self):
        """Setup the ChatHuggingFace model with error handling."""
        try:
            print(f"🔄 Attempting to initialize model: {self.model_repo}")
            
            # Create HuggingFace endpoint
            endpoint = HuggingFaceEndpoint(
                repo_id=self.model_repo,
                provider= self.provider,
                temperature=self.temperature,
                huggingfacehub_api_token=self.hf_token
            )
            
            # Create ChatHuggingFace instance
            self.llm = ChatHuggingFace(llm=endpoint, temperature=self.temperature)
            print(f"✅ Successfully initialized model: {self.model_repo}")
            
        except Exception as e:
            print(f"❌ Error setting up model: {e}")
            print("🔄 Will use fallback rule-based enhancement")
            self.llm = None

    def clean_response(self, text: str) -> str:
        # Remove everything between <think>...</think>
        return re.sub(r"<think>.*?</think>", "", text, flags=re.DOTALL).strip()

    
    def ai_enhancement(self, description: str) -> str:
        """AI-based enhancement using ChatHuggingFace."""
        # Create system and human messages
        system_message = SystemMessage(
            content="""
            You are a professional real estate copywriter. Transform basic property descriptions 
            into compelling, attractive listings that appeal to potential buyers or renters.
            Guidelines:
            - Use vivid, appealing language
            - Highlight key features and benefits
            - Add emotional appeal
            - Maintain accuracy - don't add false information
            - Keep it concise but descriptive
            - Do not include any redundant information, Only relevant details
            """)
        
        human_message = HumanMessage(content=f"Please enhance this property description: {description}")
        
        # Get response from chat model
        response = self.llm.invoke([system_message, human_message])

        response = self.clean_response(response.content.strip())

        return response

# Initialize the PropertyListingEnhancer
enhancer = PropertyListingEnhancer()
```
